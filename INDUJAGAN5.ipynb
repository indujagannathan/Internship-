```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Load the dataset
print("Loading dataset...")
df = pd.read_csv('/content/drive/MyDrive/Anime_data.csv')
print(f"Dataset loaded. Shape: {df.shape}")

# Display basic information about the dataset
print("\nDataset info:")
print(df.info())

print("\nFirst few rows of the dataset:")
print(df.head())

# Basic statistics
print("\nBasic statistics of numerical columns:")
print(df.describe())

# Ensure column names are correct
print("\nColumns in dataset:", df.columns)

# Most popular anime
print("\nTop 5 most popular anime:")
top_anime = df.sort_values('Members', ascending=False).head()
  # Assuming 'members' is the correct column name
print(top_anime[['Rating', 'Members']])  # Assuming 'title' is the correct column name

# Function to get anime recommendations
def get_recommendations(anime_title, Data, n=5):
    # Check if 'title' column exists in Data
    if 'title' in Data.columns:
        # Check if the anime_title exists in the title column
        if anime_title in Data['title'].values:
            idx = Data[Data['title'] == anime_title].index[0]

            # Calculate cosine similarity
            cosine_sim = cosine_similarity(Data[['score', 'members']])  # Assuming 'score' and 'members' are correct column names

            # Get similarity scores
            sim_scores = list(enumerate(cosine_sim[idx]))

            # Sort anime based on similarity scores
            sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

            # Get top N most similar anime
            sim_scores = sim_scores[1:n+1]
            anime_indices = [i[0] for i in sim_scores]

            # Ensure indices are within bounds
            anime_indices = [i for i in anime_indices if i < len(Data)]
            return Data['title'].iloc[anime_indices]
        else:
            print(f"'{anime_title}' not found in the dataset.")
            return None  # Return None if the title is not found
    else:
        print("'title' column not found in the dataset.")
        return None  # Return None if the column is missing

# Get recommendations for a popular anime
# Check if top_anime is not empty
if not top_anime.empty:
    popular_anime = top_anime['Title'].iloc[0]  # Use the correct casing for the column name
    print(f"\nRecommendations based on {popular_anime}:")

    # Get recommendations only if popular_anime is valid
    if popular_anime:
        recommendations = get_recommendations(popular_anime, df)
        print(recommendations)
    else:
        print("No valid popular anime title found.")
else:
    print("No popular anime found.")

print("\nAnalysis complete.")


```
